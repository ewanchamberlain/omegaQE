```
import omegaqe
from omegaqe.fields import Fields
from omegaqe.powerspectra import Powerspectra
import omegaqe.postborn as postborn
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import InterpolatedUnivariateSpline

power = Powerspectra()
```

```
plt.rcParams["figure.figsize"] = [10, 5]
plt.rcParams["xtick.major.pad"] = "8"
plt.rc("figure", facecolor="w")
```

```
LSS_fields = "kgI"
exp = "SO_goal"
fields = Fields(LSS_fields, exp=exp, N_pix_pow=10, kmax=5000)
```

```
kappa_map = fields.get_map("k", fft=False)
plt.imshow(kappa_map, cmap="jet", vmin=-1, vmax=1)
plt.colorbar()
```

```
gal_map = fields.get_map("g", fft=False)
plt.imshow(gal_map, cmap="jet", vmin=-3, vmax=3)
plt.colorbar()
```

```
ells = np.arange(30, 3001)
fiducial_kg = power.get_ps("kg", ells)
```

```
fields_kg, ks_bin, err_bin = fields.get_ps_binned(
    fields.get_map("k"), fields.get_map("g"), kmin=30, kmax=3000, nBins=50
)
```

```
plt.errorbar(
    ks_bin, fields_kg, err_bin, fmt=".", markersize=15, linewidth=3, label="Simulated"
)
plt.semilogy(ells, (2 * np.pi) ** 2 * fiducial_kg, linewidth=3, label="Fiducial")
plt.ylabel(r"$C_L^{\kappa g}$", fontsize=15)
plt.xlabel("$L$", fontsize=15)
plt.xticks(fontsize=10)
plt.yticks(fontsize=10)
plt.ylim(1e-7, 3e-5)
plt.xlim(30, 3000)
plt.legend(prop={"size": 15}, frameon=False, handlelength=1.5)
```

```
def _C_inv_splines(LSS_fields, L_max_map=5000, L_min_cut=30, L_max_cut=3000):
    C_inv = fields.fish.covariance.get_C_inv(LSS_fields, L_max_map, 353e9)
    N_fields = np.size(list(LSS_fields))
    C_inv_splines = np.empty((N_fields, N_fields), dtype=InterpolatedUnivariateSpline)
    Ls = np.arange(L_max_map + 1)
    for iii in range(N_fields):
        for jjj in range(N_fields):
            C_inv_ij = C_inv[iii, jjj]
            C_inv_ij[L_max_cut + 1 :] = 0
            C_inv_ij[:L_min_cut] = 0
            C_inv_splines[iii, jjj] = InterpolatedUnivariateSpline(Ls, C_inv_ij)
    return C_inv_splines
```

```
Ls = np.load(
    f"{omegaqe.RESULTS_DIR}/F_L_results/{LSS_fields}/{exp}/gmv/TEB/30_3000/1_2000/Ls.npy"
)
F_L_spline = InterpolatedUnivariateSpline(
    Ls,
    np.load(
        f"{omegaqe.RESULTS_DIR}/F_L_results/{LSS_fields}/{exp}/gmv/TEB/30_3000/1_2000/F_L.npy"
    ),
)
%time C_inv_spline = _C_inv_splines(LSS_fields)
```

```
omega_tem = fields.get_omega_template(
    100, F_L_spline=F_L_spline, C_inv_spline=C_inv_spline, tracer_noise=True
)
```

```
omega_Ls, omega_ps_camb = power.cosmo.get_postborn_omega_ps(ellmax=8000)
omega_ps = postborn.omega_ps(omega_Ls)
omega_ps_spline = InterpolatedUnivariateSpline(omega_Ls, omega_ps)
```

```
wiener = F_L_spline(fields.kM)
omega_tem_filt = omega_tem * wiener
plt.imshow(
    np.fft.irfft2(omega_tem_filt, norm="forward"), cmap="jet", vmin=-0.01, vmax=0.01
)
plt.colorbar()
```

```
omega_tem_ps, om_ks_bin, om_err_bin = fields.get_ps_binned(
    omega_tem, kmin=30, kmax=3000, nBins=50
)
plt.errorbar(
    om_ks_bin,
    omega_tem_ps,
    om_err_bin,
    fmt=".",
    markersize=15,
    linewidth=3,
    label=r"Simulated $C^{\omega^{\mathrm{tem}}}_L$",
)
plt.semilogy(
    omega_Ls, (2 * np.pi) ** 2 * omega_ps, linewidth=5, label=r"Fiducial $C^{\omega}_L$"
)
plt.semilogy(
    ells,
    (2 * np.pi) ** 2 * omega_ps_spline(ells) / F_L_spline(ells),
    linewidth=5,
    label=r"$C^{\omega}F_L^{-1}$",
)
plt.ylabel("$C_L$", fontsize=15)
plt.xlabel("$L$", fontsize=15)
plt.xticks(fontsize=10)
plt.yticks(fontsize=10)
plt.ylim(6e-11, 1e-9)
plt.xlim(30, 3000)
plt.legend(prop={"size": 15}, frameon=False, handlelength=1.5)
```
