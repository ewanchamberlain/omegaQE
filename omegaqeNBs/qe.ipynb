```
import omegaqe
import numpy as np
from omegaqe.noise import Noise
from scipy.interpolate import InterpolatedUnivariateSpline
import matplotlib.pyplot as plt
from omegaqe.qe import QE
import pandas as pd

plt.rcParams["figure.figsize"] = [10, 5]
plt.rc("figure", facecolor="w")
```

```
%time _qe = QE(exp="S4")
_noise = Noise()
%time _noise.setup_cmb_noise()
```

```
%time _qe_base = QE(exp="S4_base")
```

```
Ls = np.arange(1, 4001)
plt.semilogy(
    Ls,
    (Ls * (Ls + 1)) * _qe.cmb["TT"].lenCl_spline(Ls) / (2 * np.pi),
    label=r"$C_{\ell}^{TT}$",
)
plt.semilogy(
    Ls,
    (Ls * (Ls + 1)) * _qe.cmb["TT"].gradCl_spline(Ls) / (2 * np.pi),
    label=r"$C_{\ell}^{T\Delta T}$",
)
plt.semilogy(
    Ls[10:],
    (Ls[10:] * (Ls[10:] + 1)) * _qe.cmb["TT"].N_spline(Ls[10:]) / (2 * np.pi),
    label=r"$N_{\ell}^{\mathrm{SO}}$",
)
N = _noise.get_cmb_gaussian_N("TT", None, None, exp="S4_base")[:6001]
Ls = np.arange(np.size(N))
plt.semilogy(
    Ls, (Ls * (Ls + 1)) * N / (2 * np.pi), label=r"$N_{\ell}^{\mathrm{SO}}+\mathrm{fg}$"
)
plt.semilogy(
    Ls,
    (Ls * (Ls + 1)) * _qe_base.cmb["TT"].N_spline(Ls) / (2 * np.pi),
    label=r"$N_{\ell}^{\mathrm{SO}}+\mathrm{fg}$ (spline)",
)
plt.ylim(1e-14, 1e-9)
plt.xlim(0, 6000)
plt.ylabel(r"$\ell(\ell+1)C_{\ell}/2\pi$", fontsize=14)
plt.xlabel(r"$\ell$", fontsize=14)
plt.legend(prop={"size": 12})
```

```
def get_N0(N0_file, typ):
    dict = {
        "TT": (1, 8),
        "TE": (2, 9),
        "EE": (3, 10),
        "TB": (4, 11),
        "EB": (5, 12),
        "Pol": (6, 13),
        "MV": (7, 14),
    }
    phi_index, curl_index = dict[typ]
    df = pd.read_csv(N0_file, sep=r"\s+", header=None)
    return np.array(df[phi_index]), np.array(df[curl_index])
```
